# Guía: Cómo leer y entender los errores en Python

Cuando tu código falla, Python imprime un mensaje de error llamado **Traceback**. Aunque al principio puede parecer confuso, es tu mejor herramienta para corregir el programa.

## Anatomía de un Error

Para leer un Traceback de manera efectiva, sigue estos pasos:
1. **Mira la última línea:** Aquí Python te dice exactamente qué tipo de error ocurrió y te da una breve descripción del problema.
2. **Mira las líneas anteriores:** Python te indicará en qué archivo y en qué línea específica de tu código ocurrió el fallo (por ejemplo, `line 14`).
3. **Revisa el contexto:** A veces el error real está en la línea inmediatamente anterior a la que indica Python.

---

## Errores más comunes

### 1. `SyntaxError` (Error de Sintaxis)
**Qué significa:** Escribiste algo que Python no puede entender porque rompe las reglas de escritura del lenguaje.
**Causas comunes:**
* Olvidar los dos puntos (`:`) al final de un `if`, `for`, `while` o `def`.
* Escribir mal una palabra clave (por ejemplo, `whle` en lugar de `while`).

### 2. `IndentationError` (Error de Indentación)
**Qué significa:** Hay un problema con los espacios al inicio de la línea. Python usa la indentación para saber qué código pertenece a qué bloque.
**Causas comunes:**
* Olvidar indentar el código debajo de un `if`, `for` o función.
* Mezclar espacios y tabulaciones en el mismo documento.
* Indentar una línea que no debería estar indentada.

### 3. `NameError` (Error de Nombre)
**Qué significa:** Intentaste usar una variable o función que Python no reconoce.
**Causas comunes:**
* Escribir mal el nombre de una variable (ej. creaste `contador` pero escribiste `cantador`).
* Intentar usar una variable antes de haberle asignado un valor.
* Olvidar poner comillas en un texto, haciendo que Python crea que es una variable.

### 4. `TypeError` (Error de Tipo)
**Qué significa:** Intentaste hacer una operación con tipos de datos incompatibles.
**Causas comunes:**
* Intentar sumar un número y un texto (ej. `5 + "hola"`).
* Intentar usar una función matemática en un texto que no ha sido convertido a entero o flotante.

### 5. `EOFError` o `unexpected EOF while parsing` (Fin de Archivo)
**Qué significa:** Python llegó al final de tu código (End Of File) pero esperaba algo más para poder terminar de leer la instrucción.
**Causas comunes:**
* **Falta de cierre:** Abriste un paréntesis `(`, corchete `[` o comillas `"`, pero olvidaste cerrarlos. Revisa la línea del error y la anterior.
* **Lectura de datos (`EOFError: EOF when reading a line`):** Muy común en plataformas de evaluación automática. Ocurre si usas un `input()` extra que pide datos cuando el sistema ya no tiene más información para entregarte, o si tienes un `input()` atrapado en un ciclo infinito.

## STR-01 · `len(s)` y `s[x]`

Dos operaciones fundamentales sobre un string:

- `len(s)` devuelve la cantidad de caracteres del string.
- `s[x]` devuelve el carácter en la posición `x`.

Los índices empiezan en **0**: la primera letra es `s[0]`, y la última es `s[len(s) - 1]`. Si pedimos un índice fuera del rango, Python lanza un error.

También podemos usar índices negativos: `s[-1]` es la última letra, `s[-2]` la penúltima, y así.

````python
# Ejemplo: medir y acceder
palabra = "Python"

print("Longitud:", len(palabra))
print("Primera letra:", palabra[0])
print("Última letra:", palabra[len(palabra) - 1])
print("Última letra (con índice negativo):", palabra[-1])
````

## STR-02 · Iterar carácter a carácter

Cuando queremos revisar todos los caracteres de un string, hay tres formas de hacerlo. Las tres producen el mismo recorrido; cambia lo que la variable representa.

### Forma 1 · `for c in s`

La más simple y directa. En cada vuelta, `c` toma el valor de un carácter.

````python
palabra = "Hola"

for c in palabra:
    print(c)
````

````python
for i in "jajajajaja":
  print(i)
````

### Forma 2 · `for i in range(len(s))`

Recorre los índices en lugar de los caracteres. Útil cuando necesitamos saber la posición.

````python
palabra = "Hola"

for i in range(len(palabra)):
    print(i, palabra[i])
````

### Forma 3 · `while`

La más flexible. Nos deja decidir cuándo parar. Requiere las tres piezas de siempre: valor inicial, condición y actualización.

````python
palabra = "Hola"
i = 0

while i < len(palabra):
    print(palabra[i])
    i = i + 1
````

### ¿Cuál usar?

| Forma | Cuándo |
|---|---|
| `for c in s` | Sólo importa el carácter |
| `for i in range(len(s))` | Necesitamos la posición |
| `while` | Recorrido flexible o parada anticipada |

## STR-03 · Substring contenido en otro (`in`)

El operador `in` responde a la pregunta *"¿este substring está dentro del string?"*. Devuelve `True` o `False`, así que se puede usar directamente en un `if` o en un `while`.

Es sensible a mayúsculas: `"A" in "casa"` es `False`, `"a" in "casa"` es `True`.

````python
# Ejemplos rápidos
print("a" in "casa")          # True
print("z" in "casa")          # False
print("as" in "casa")         # True
print("CASA" in "casa")       # False (sensible a mayúsculas)
print("gato" in "tengo un gato negro")   # True
````

````python
1 in [1, 2, 3]
````

El operador `in` también nos permite escribir el contador de vocales de forma mucho más corta:

````python
# Contar vocales usando in
palabra = input("Ingresa una palabra: ")
vocales = 0

for c in palabra:
    if c in "aeiou":
        vocales = vocales + 1

print("La palabra tiene", vocales, "vocales")
````

## Tipo de Secuencia de Texto — `str`

Los datos de texto en Python se manejan con objetos `str`, o cadenas (*strings*). Las cadenas son secuencias inmutables de puntos de código Unicode. Las literales de cadena se escriben de varias maneras:

* **Comillas simples:** `'permite comillas "dobles" incrustadas'`
* **Comillas dobles:** `"permite comillas 'simples' incrustadas"`
* **Comillas triples:** `'''Tres comillas simples'''`, `"""Tres comillas dobles"""`

Las cadenas de comillas triples pueden extenderse a lo largo de varias líneas; todos los espacios en blanco asociados se incluirán en la literal de la cadena.

Las cadenas de texto también se pueden crear a partir de otros objetos utilizando el constructor `str`.

### `class str(object='')`
### `class str(object=b'', encoding='utf-8', errors='strict')`

Retorna una versión en cadena de texto del objeto (*object*). Si no se proporciona un objeto, devuelve la cadena vacía `''`. De lo contrario, el comportamiento de `str()` depende de si se proporciona el parámetro *encoding* o *errors*.

---

### Métodos de Cadena (String Methods)

Las cadenas implementan todas las operaciones comunes de las secuencias, además de los siguientes métodos adicionales:

**`str.capitalize()`**
Retorna una copia de la cadena con su primer carácter en mayúscula y el resto en minúsculas.

**`str.lower()`**
Retorna una copia de la cadena con todos los caracteres convertidos a minúsculas.

**`str.upper()`**
Retorna una copia de la cadena con todos los caracteres convertidos a mayúsculas.

**`str.replace(old, new[, count])`**
Retorna una copia de la cadena con todas las apariciones de la subcadena *old* reemplazadas por *new*. Si se proporciona el argumento opcional *count*, solo se reemplazan las primeras *count* apariciones.

**`str.split(sep=None, maxsplit=-1)`**
Retorna una lista de las palabras en la cadena, usando *sep* como la cadena delimitadora. Si se especifica *maxsplit*, se realiza como máximo ese número de divisiones (dejando el resto de la cadena como el último elemento de la lista).

**`str.join(iterable)`**
Retorna una cadena que es la concatenación de las cadenas en el *iterable*. Se genera un `TypeError` si hay valores que no sean cadenas en el iterable. El separador entre los elementos es la cadena que proporciona este método.

**`str.strip([chars])`**
Retorna una copia de la cadena con los caracteres iniciales y finales eliminados. El argumento *chars* es una cadena que especifica el conjunto de caracteres a eliminar. Si se omite o es `None`, por defecto se eliminan los espacios en blanco.
