````python
# Ejemplo de ordenamiento de enteros sin usar el método .sort()
# Implementaremos el algoritmo de ordenamiento por selección (Selection Sort)

def ordenar_enteros(lista):
    n = len(lista)
    # Recorremos cada elemento de la lista
    for i in range(n):
        # Suponemos que el primer elemento no ordenado es el mínimo
        indice_minimo = i
        # Buscamos el menor elemento en el resto de la lista
        for j in range(i + 1, n):
            if lista[j] < lista[indice_minimo]:
                indice_minimo = j
        # Intercambiamos el mínimo encontrado con el elemento de la posición i
        aux = lista[i]
        lista[i] = lista[indice_minimo]
        lista[indice_minimo] = aux

# Prueba del algoritmo
numeros = [64, 25, 12, 22, 11]
print("Lista original:", numeros)

ordenar_enteros(numeros)
print("Lista ordenada:", numeros)
````

## ORD-03 · Criterio propio con `key=`

Cuando queremos ordenar por un criterio que Python no deduce solo, le pasamos una función a `key`. Esa función recibe UN elemento y devuelve el valor por el que queremos ordenar.

````python
# Ordenar strings por largo en vez de alfabéticamente
palabras = ["programacion", "hola", "gato", "universidad"]

def largo(s):
    return len(s)

palabras.sort(key=largo)
print(palabras)
````

````python
palabras2 = ["bcd", "abcd", "hgfd", "asdfyt"]
palabras2.sort()
print(palabras2)
````

````python
palabras2.sort(key=largo)
print(palabras2)
````

````python
palabras2.sort(key=largo, reverse=True)
print(palabras2)
````

**Observa:** le pasamos `key=largo` (sin paréntesis). Le damos la función misma, no la llamamos. Python se encargará de llamarla por cada elemento.

### Ordenar lista de objetos

````python
class Estudiante:
    def __init__(self, nombre, promedio):
        self.nombre = nombre
        self.promedio = promedio

    def __str__(self):
        return self.nombre + " (" + str(self.promedio) + ")"

estudiantes = [
    Estudiante("Ana", 6.5),
    Estudiante("Luis", 4.0),
    Estudiante("María", 5.8),
    Estudiante("Pedro", 3.2)
]

def obtener_promedio(e):
    return e.promedio

estudiantes.sort(key=obtener_promedio)

for e in estudiantes:
    print(e)
````

## ORD-05 · Algoritmo de selección

¿Cómo ordena Python por dentro? Vamos a implementar nosotros mismos un algoritmo de ordenamiento para entender la lógica.

### Idea del algoritmo de selección

1. Recorrer la lista desde el inicio.
2. Para cada posición, buscar el elemento más chico entre los que vienen después.
3. Intercambiar ese mínimo con el elemento de la posición actual.
4. Repetir hasta llegar al final.

````python
def ordenamiento_seleccion(lista):
    for i in range(len(lista)):
        # buscar el índice del mínimo desde i hasta el final
        indice_min = i
        for j in range(i + 1, len(lista)):
            if lista[j] < lista[indice_min]:
                indice_min = j
        # intercambiar lista[i] con lista[indice_min]
        aux = lista[i]
        lista[i] = lista[indice_min]
        lista[indice_min] = aux

# Prueba
numeros = [5, 2, 8, 1, 4]
ordenamiento_seleccion(numeros)
print(numeros)
````

## ORD-06 · Algoritmos más eficientes

El algoritmo de selección es fácil de entender pero **lento** cuando las listas son grandes:

- Con 10 elementos: ~45 comparaciones.
- Con 100 elementos: ~5000 comparaciones.
- Con 1000 elementos: ~500 000 comparaciones.

Si la lista crece 10 veces, el trabajo crece 100 veces. Esto se llama crecimiento cuadrático.

**Existen algoritmos más eficientes:**

- **Merge Sort** y **Quick Sort** dividen el problema en mitades (divide y vencerás). Son mucho más rápidos para listas grandes.
- Python usa **Timsort** internamente, una combinación optimizada de varios algoritmos.
- En cursos avanzados (IIC2133) estudiarán estos algoritmos con detalle.

````python
import time
import random
import sys

# Aumentar límite de recursión por seguridad con listas grandes en Quick Sort
sys.setrecursionlimit(20000000)

# 1. Algoritmo de Selección (Selection Sort - Sin usar .sort())
def selection_sort(lista):
    arr = lista.copy()
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            if arr[j] < arr[min_idx]:
                min_idx = j
        arr[i], arr[min_idx] = arr[min_idx], arr[i]
    return arr

# 2. Merge Sort (Divide y Vencerás)
def merge_sort(lista):
    if len(lista) <= 1:
        return lista

    # Paso de división
    medio = len(lista) // 2
    izquierda = merge_sort(lista[:medio])
    derecha = merge_sort(lista[medio:])

    # Paso de mezcla (merge)
    return merge(izquierda, derecha)

def merge(izq, der):
    resultado = []
    i = j = 0
    while i < len(izq) and j < len(der):
        if izq[i] < der[j]:
            resultado.append(izq[i])
            i += 1
        else:
            resultado.append(der[j])
            j += 1
    resultado.extend(izq[i:])
    resultado.extend(der[j:])
    return resultado

# 3. Quick Sort (Divide y Vencerás)
def quick_sort(lista):
    if len(lista) <= 1:
        return lista
    else:
        pivote = lista[len(lista) // 2]
        izq = [x for x in lista if x < pivote]
        medio = [x for x in lista if x == pivote]
        der = [x for x in lista if x > pivote]
        return quick_sort(izq) + medio + quick_sort(der)

# Generación de datos de prueba
tamano = 2000
numeros_prueba = [random.randint(1, 10000) for _ in range(tamano)]
print(f"--- Comparación de algoritmos con {tamano} enteros ---\n")

# Medición de Selection Sort
t_inicio = time.time()
_ = selection_sort(numeros_prueba)
t_fin = time.time()
t_selection = t_fin - t_inicio
print(f"Selection Sort (Sin sort nativo): {t_selection:.6f} segundos")

# Medición de Merge Sort
t_inicio = time.time()
_ = merge_sort(numeros_prueba)
t_fin = time.time()
t_merge = t_fin - t_inicio
print(f"Merge Sort (Divide y Vencerás):  {t_merge:.6f} segundos")

# Medición de Quick Sort
t_inicio = time.time()
_ = quick_sort(numeros_prueba)
t_fin = time.time()
t_quick = t_fin - t_inicio
print(f"Quick Sort (Divide y Vencerás):  {t_quick:.6f} segundos")

# Medición de sorted() (Timsort de Python - crea copia)
t_inicio = time.time()
_ = sorted(numeros_prueba)
t_fin = time.time()
t_sorted = t_fin - t_inicio
print(f"Método sorted() de Python:        {t_sorted:.6f} segundos")

# Medición de .sort() (Timsort de Python - in-place)
copia_prueba = numeros_prueba.copy()
t_inicio = time.time()
copia_prueba.sort()
t_fin = time.time()
t_sort_metodo = t_fin - t_inicio
print(f"Método .sort() de Python (in-place): {t_sort_metodo:.6f} segundos")
````
