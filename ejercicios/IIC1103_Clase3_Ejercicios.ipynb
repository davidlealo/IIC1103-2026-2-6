## 1. Operaciones con booleanos

Podemos usar: `and`, `or` y `not`

```python
edad = int(input("Edad: "))
tiene_carnet = bool(input("Tiene carnet? (True/False): "))
# ¿Puede conducir? (mayor de edad Y con carnet)
(edad >= 18) and (tiene_carnet == True)
```

## 2. Comparaciones entre strings


Para poder comparar strings podemos usar: `==` y `!=`

```python
"hola" == "hola"      # True
"hola" == "chao"      # False
"hola" != "chao"      # True

# Ojo: distinguen mayúsculas y espacios
"hola" == "Hola"      # False
"hola" == "hola "     # False   ← espacio al final
```

## 3. Variables

## Reglas para nombrar variables en Python

Al programar en Python, los nombres de las variables deben cumplir con un conjunto de reglas estrictas definidas por el lenguaje. Si no se respetan, el intérprete arrojará un error de sintaxis (SyntaxError).

### Reglas Obligatorias

1. **Caracteres permitidos:**
   Los nombres de las variables solo pueden contener letras (a-z, A-Z), números (0-9) y guiones bajos (_). No se permiten espacios en blanco ni caracteres especiales (como @, #, $, %, -, etc.).

2. **Inicio del nombre de la variable:**
   Un nombre de variable debe comenzar obligatoriamente con una letra o con un guion bajo (_).
   - Correcto: `variable`, `_variable`, `variable1`
   - Incorrecto: `1variable` (no puede comenzar con un número).

3. **Distinción entre mayúsculas y minúsculas (Case-sensitive):**
   Python distingue entre letras mayúsculas y minúsculas. Por lo tanto, `mivariable`, `MiVariable` y `MIVARIABLE` son tres variables completamente diferentes.

4. **Palabras reservadas (Keywords):**
   No se pueden usar las palabras reservadas del lenguaje como nombres de variables. Estas son palabras que Python usa para su propia sintaxis y estructura.
   - Ejemplos de palabras prohibidas: `False`, `None`, `True`, `and`, `as`, `assert`, `break`, `class`, `continue`, `def`, `del`, `elif`, `else`, `except`, `finally`, `for`, `from`, `global`, `if`, `import`, `in`, `is`, `lambda`, `nonlocal`, `not`, `or`, `pass`, `raise`, `return`, `try`, `while`, `with`, `yield`.

### Convenciones y Buenas Prácticas (PEP 8)

Aunque no son reglas estrictas que causen errores si se omiten, la comunidad de Python sigue el estándar PEP 8 para mantener el código legible:

- **Formato snake_case:** Se recomienda escribir los nombres de las variables en letras minúsculas, separando las palabras con guiones bajos para mejorar la legibilidad.
  - Ejemplo recomendado: `contador_de_visitas`, `nombre_usuario`.
- **Nombres descriptivos:** El nombre de la variable debe indicar claramente qué tipo de dato o información almacena.
  - Evitar: `x`, `a`, `val` (a menos que sean iteradores cortos).
  - Preferir: `total_ventas`, `edad_cliente`.
- **Uso de un solo guion bajo inicial:** Un nombre que comienza con un guion bajo (por ejemplo, `_variable_interna`) indica por convención que esa variable es de uso interno y no debería ser accedida directamente desde fuera del módulo o clase, aunque el lenguaje no lo impida técnicamente.


Para asignar variables usamos `=` para comparar `==` para incrementar `=+`


```python
# Forma larga
puntos = 0
puntos = puntos + 10    # ahora puntos vale 10

# Forma corta (equivalente)
puntos = 0
puntos += 10            # ahora puntos vale 10

# También funciona con otras operaciones
puntos -= 5   # resta 5
puntos *= 2   # multiplica por 2
puntos /= 4   # divide por 4
```

## 4. print(), input() y convertir strings

Estos elementos los vimos la clase anterior

```python
# 1. Un valor
print("Hola mundo")
print(42)

# 2. Varios valores separados por coma (agrega espacios)
edad = 20
print("Edad:", edad, "años")
# Salida: Edad: 20 años

# 3. Concatenando strings con + (sin espacios automáticos)
print("Edad: " + str(edad) + " años")
# Salida: Edad: 20 años
```
