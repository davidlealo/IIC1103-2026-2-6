## OBJ-10 · Un objeto como atributo de otro

Un atributo puede guardar un objeto entero. Para acceder a sus datos internos se usa doble punto: `objeto.atributo.sub_atributo`.

```python
class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.puntaje = 0

    def sumar_puntos(self, n):
        self.puntaje = self.puntaje + n

class Partida:
    def __init__(self, jugador):
        self.jugador = jugador    # un Jugador como atributo
        self.turno = 1

    def jugar_ronda(self):
        self.jugador.sumar_puntos(10)
        self.turno = self.turno + 1

# Uso
ana = Jugador("Ana")
p = Partida(ana)

print(p.jugador.nombre)         # Ana (doble punto)
print(p.jugador.puntaje)        # 0

p.jugar_ronda()
p.jugar_ronda()

print(p.jugador.puntaje)        # 20
print(p.turno)                  # 3
```

## OBJ-11 · Lista de objetos como atributo

Un atributo puede ser una lista que contiene objetos. Este es el patrón más potente de POO.

```python
class Mazo:
    def __init__(self):
        self.cartas = []                 # lista vacía de Carta

    def agregar(self, carta):
        self.cartas.append(carta)

    def cantidad(self):
        return len(self.cartas)

    def contar_pinta(self, pinta):
        total = 0
        for c in self.cartas:            # recorrer la lista interna
            if c.pinta == pinta:
                total = total + 1
        return total

    def __str__(self):
        return "Mazo con " + str(self.cantidad()) + " cartas"

# Uso
m = Mazo()
m.agregar(Carta(1, "corazones"))
m.agregar(Carta(13, "picas"))
m.agregar(Carta(5, "corazones"))

print(m)
print("Cartas de corazones:", m.contar_pinta("corazones"))
```

## OBJ-12 · Crear objeto desde otro constructor

Dentro del `__init__` de una clase podemos crear objetos de otras clases. Esto permite que un objeto se inicialice completo, sin tener que armarlo pieza por pieza desde afuera.

```python
class MazoCompleto:
    def __init__(self):
        self.cartas = []
        pintas = ["corazones", "picas", "tréboles", "diamantes"]
        for p in pintas:
            for n in range(1, 14):
                self.cartas.append(Carta(n, p))    # crear Carta acá

    def cantidad(self):
        return len(self.cartas)

m = MazoCompleto()
print("Cartas totales:", m.cantidad())      # 52
print("Primera carta:", m.cartas[0])         # 1 de corazones
print("Carta 25:", m.cartas[25])             # 13 de picas
```

**Ejercicio 5.** Define una clase `BibliotecaInicial` cuyo constructor reciba una lista de títulos y cree automáticamente un objeto `Libro` para cada título.

Usa esta clase `Libro`:

```python
class Libro:
    def __init__(self, titulo):
        self.titulo = titulo
        self.disponible = True

    def __str__(self):
        return self.titulo + (" (disponible)" if self.disponible else " (prestado)")

# tu código: define BibliotecaInicial
class BibliotecaInicial:
    def __init__(self, titulos):
        self.libros = []
        for t in titulos:
            self.libros.append(Libro(t))

# Prueba
b = BibliotecaInicial(["El Quijote", "Hamlet", "1984", "Rayuela"])
for libro in b.libros:
    print(libro)
```

## Ejemplo integrador: sistema de Partida completo

Este sistema usa los 4 conceptos de hoy. Léelo con calma y ejecútalo.

```python
import random

class Mazo:
    def __init__(self):
        self.cartas = []
        for p in ["corazones", "picas", "tréboles", "diamantes"]:
            for n in range(1, 14):
                self.cartas.append(Carta(n, p))

    def barajar(self):
        random.shuffle(self.cartas)

    def repartir(self):
        return self.cartas.pop()

    def cantidad(self):
        return len(self.cartas)

class Jugador:
    def __init__(self, nombre):
        self.nombre = nombre
        self.mano = []

    def recibir(self, carta):
        self.mano.append(carta)

    def mostrar_mano(self):
        print(self.nombre, "tiene:")
        for c in self.mano:
            print("  -", c)

class Partida:
    def __init__(self, nombre_jugador):
        self.jugador = Jugador(nombre_jugador)
        self.mazo = Mazo()
        self.mazo.barajar()

    def repartir_mano(self, n):
        for i in range(n):
            carta = self.mazo.repartir()
            self.jugador.recibir(carta)

# Uso
p = Partida("Ana")
p.repartir_mano(5)
p.jugador.mostrar_mano()
print("Quedan", p.mazo.cantidad(), "cartas en el mazo")
```

Fíjate en lo que acabas de ver:

- `Partida` contiene a `Jugador` y `Mazo` (OBJ-10).
- `Jugador` contiene una lista de `Carta` (OBJ-11).
- `Mazo` contiene una lista de `Carta` (OBJ-11).
- `Partida.__init__` crea `Jugador` y `Mazo` adentro (OBJ-12).
- Para recorrer `self.mano` usamos un `for` (OBJ-09).

## Desafío · clase Curso

Define una clase `Curso` que modele un curso completo:

1. `__init__(self, nombre_curso)`: recibe el nombre y crea una lista vacía de estudiantes.
2. `inscribir(self, estudiante)`: agrega un `Estudiante` a la lista.
3. `cantidad_inscritos(self)`: devuelve cuántos estudiantes hay.
4. `promedio_curso(self)`: devuelve el promedio de los promedios de todos los estudiantes.
5. `mejor_estudiante(self)`: devuelve el `Estudiante` con mayor promedio.
6. `aprobados(self)`: devuelve una lista con los estudiantes cuyo promedio es ≥ 4.0.
7. `__str__(self)`: devuelve algo como `"Curso IIC1103 con 30 inscritos"`.

```python
class Curso:
    def __init__(self, nombre_curso):
        self.nombre_curso = nombre_curso
        self.estudiantes = []

    def inscribir(self, estudiante):
        self.estudiantes.append(estudiante)

    def cantidad_inscritos(self):
        return len(self.estudiantes)

    def promedio_curso(self):
        if len(self.estudiantes) == 0:
            return 0
        suma = 0
        for e in self.estudiantes:
            suma += e.promedio()
        return suma / len(self.estudiantes)

    def mejor_estudiante(self):
        if len(self.estudiantes) == 0:
            return None
        mejor = self.estudiantes[0]
        for e in self.estudiantes:
            if e.promedio() > mejor.promedio():
                mejor = e
        return mejor

    def aprobados(self):
        lista_aprobados = []
        for e in self.estudiantes:
            if e.promedio() >= 4.0:
                lista_aprobados.append(e)
        return lista_aprobados

    def __str__(self):
        return f"Curso {self.nombre_curso} con {self.cantidad_inscritos()} inscritos"

# Pruebas
iic1103 = Curso("IIC1103")

ana = Estudiante("Ana", "Ingeniería")
ana.agregar_nota(6.5)
ana.agregar_nota(7.0)

luis = Estudiante("Luis", "Letras")
luis.agregar_nota(3.5)
luis.agregar_nota(3.8)

maria = Estudiante("María", "Ingeniería")
maria.agregar_nota(5.0)
maria.agregar_nota(5.5)

iic1103.inscribir(ana)
iic1103.inscribir(luis)
iic1103.inscribir(maria)

print(iic1103)                            # Curso IIC1103 con 3 inscritos
print("Promedio del curso:", iic1103.promedio_curso())
print("Mejor:", iic1103.mejor_estudiante().nombre)
print("Aprobados:")
for e in iic1103.aprobados():
    print("  -", e.nombre)
```
