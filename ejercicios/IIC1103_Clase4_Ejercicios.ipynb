## FLU-01 · `if` – `else`

Sirve para decidir entre dos caminos posibles. Si la condición es `True` se ejecuta el primer bloque; si es `False`, el segundo.

```
if condicion:
    # bloque 1
else:
    # bloque 2
```

Los dos puntos y la indentación son parte del lenguaje. Un solo espacio de más o de menos rompe el programa.

````python
# Ejemplo: par o impar
numero = int(input("Ingresa un número entero: "))

if numero % 2 == 0:
    print("El número es par")
else:
    print("El número es impar")
````

## FLU-02 · `if` – `elif` – `else`

Cuando hay más de dos casos posibles, agregamos uno o más `elif` entre el `if` y el `else`.

```
if condicion_1:
    # bloque 1
elif condicion_2:
    # bloque 2
elif condicion_3:
    # bloque 3
else:
    # bloque por defecto
```

Python evalúa las condiciones en orden y ejecuta el primer bloque cuya condición sea `True`. El resto se salta por completo.

````python
# Ejemplo: clasificación de nota
nota = float(input("Ingresa tu nota: "))

if nota >= 6.0:
    print("Excelente")
elif nota >= 5.0:
    print("Muy bueno")
elif nota >= 4.0:
    print("Aprobado")
else:
    print("Reprobado")
````

## FLU-03 · `if` sin `else`

No siempre hace falta un `else`. A veces sólo queremos hacer algo si la condición se cumple, y no hacer nada en caso contrario.

```
if condicion:
    # bloque

# el programa continúa aquí en ambos casos
```

Un `if` sin `else` es más simple y más legible cuando el caso `False` no requiere ninguna acción.

````python
# Ejemplo: descuento por edad
edad = int(input("Ingresa tu edad: "))
precio = 5000

if edad >= 65:
    precio = precio * 0.7   # 30% de descuento

print("Total a pagar: $", precio)
````

También podemos encadenar varios `if` – `elif` sin un `else` final: si ninguna condición se cumple, el programa simplemente no hace nada.

````python
# Ejemplo: advertencias de temperatura
temp = float(input("Temperatura: "))

if temp < 0:
    print("Advertencia: congelamiento")
elif temp > 35:
    print("Advertencia: calor extremo")

print("Fin del programa")
````

## FLU-04 · `if` anidados

Podemos poner un `if` dentro de otro cuando una decisión depende de otra decisión previa. La indentación indica el nivel de anidación.

```
if condicion_externa:
    if condicion_interna:
        # bloque interno True
    else:
        # bloque interno False
else:
    # bloque externo False
```

El `if` interno sólo se evalúa cuando la condición externa dio `True`. Cuando anidamos muchos niveles el código se vuelve difícil de leer: en varios casos se puede reemplazar por una condición combinada con `and`.

````python
# Ejemplo: ¿puede conducir?
edad = int(input("Edad: "))
licencia = input("¿Tienes licencia? (si/no): ")

if edad >= 18:
    if licencia == "si":
        print("Puedes conducir")
    else:
        print("Debes obtener licencia")
else:
    print("Aún no puedes conducir")
````
