## LIS-10 · Listas de listas rectangulares

Una lista rectangular es una lista donde cada elemento es a su vez una lista del mismo tamaño. Se visualiza como una tabla.

Para acceder a un elemento se usan dos índices seguidos: `p[i][j]`, donde `i` es la fila y `j` la columna.

````python
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

print(matriz[0])        # [1, 2, 3]   (toda la primera fila)
print(matriz[1][2])     # 6           (fila 1, columna 2)
print(matriz[2][0])     # 7

print(len(matriz))      # 3   (cantidad de filas)
print(len(matriz[0]))   # 3   (cantidad de columnas)
````

Para recorrer una matriz completa usamos ciclos anidados (igual que vimos en FLU-07).

````python
# Recorrer y sumar todos los elementos
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

total = 0
for fila in matriz:
    for valor in fila:
        total = total + valor

print("Suma total:", total)   # 45
````

## LIS-11 · Listas de listas de distinto tamaño

Las sublistas no tienen por qué ser todas del mismo largo. En muchos casos reales cada una es independiente. Para recorrer usamos `len(cursos[i])` en lugar de un largo común.

````python
cursos = [
    ["Ana", "Luis", "María"],                # 3 estudiantes
    ["Pedro", "Diego"],                      # 2 estudiantes
    ["Camila", "Julia", "Sofía", "Marco"]    # 4 estudiantes
]

for i in range(len(cursos)):
    print("Curso", i, "tiene", len(cursos[i]), "estudiantes:")
    for nombre in cursos[i]:
        print("  -", nombre)
````

## LIS-12 · Cuidado: `B = A` no es una copia

Cuando asignamos `B = A`, no estamos creando una lista nueva: `B` y `A` apuntan a la misma lista en memoria. Modificar una afecta a la otra.

````python
# El problema: B = A crea un alias, no una copia
A = [1, 2, 3]
B = A              # B apunta al mismo lugar que A

B[0] = 99          # modificamos a través de B

print("A =", A)    # [99, 2, 3]   ¡A también cambió!
print("B =", B)    # [99, 2, 3]
````

````python
A = [1, 2, 3]
B = A  # No es una copia, es una referencia

print(id(A))  # Ejemplo: 140353123456
print(id(B))  # Ejemplo: 140353123456 (¡Es el mismo número!)
````

````python
A = [1, 2, 3]
B = A

print(A is B)  # Devuelve: True
````

````python
A = [1, 2, 3]
B = A

B.append(4)

print(A)  # Resultado: [1, 2, 3, 4] (¡A también cambió!)
print(B)  # Resultado: [1, 2, 3, 4]
````

````python
# La solución: usar slicing [:] o la función list()
A = [1, 2, 3]
B = A[:]           # copia independiente
# también vale: B = list(A)

B[0] = 99

print("A =", A)    # [1, 2, 3]   A no cambió
print("B =", B)    # [99, 2, 3]
````

````python
A = [1, 2, 3]
B = A.copy()  # Copia real superficial

B.append(4)

print(A)  # [1, 2, 3]    <- El original se queda intacto
print(B)  # [1, 2, 3, 4] <- Solo B cambia
print(A is B)  # False   <- Ya no apuntan al mismo lugar
````

````python
import copy

A = [[1, 2], [3, 4]]

# ERROR Intento con .copy() (Falla en estructuras complejas)
B = A.copy()
B[0].append(99)
print(A)  # [[1, 2, 99], [3, 4]] <- ¡A también cambió por dentro!

# Solución con deepcopy() (Copia real absoluta)
A = [[1, 2], [3, 4]]
C = copy.deepcopy(A)
C[0].append(99)

print(A)  # [[1, 2], [3, 4]]    <- El original ahora sí está seguro
print(C)  # [[1, 2, 99], [3, 4]] <- Solo el clon profundo cambió
````

## LIS-13 · Conversión: string → lista con `.split()`

El método `.split()` parte un string en una lista de substrings:

- **Sin argumento:** separa por espacios en blanco.
- **Con argumento:** separa usando ese carácter como delimitador.

El resultado siempre es una lista de strings, aunque el contenido se vea como números.

````python
# Sin argumento: separa por espacios
frase = "Hola mundo Python"
palabras = frase.split()
print(palabras)         # ['Hola', 'mundo', 'Python']
print(len(palabras))    # 3

# Con argumento: separa por el carácter indicado
csv = "Ana,Luis,María,Pedro"
nombres = csv.split(",")
print(nombres)          # ['Ana', 'Luis', 'María', 'Pedro']

fecha = "2026-10-13"
partes = fecha.split("-")
print(partes)           # ['2026', '10', '13']
# Nota: son strings, no enteros
````
