## LIS-05 · Cortar: `p[i:j]`

El slicing funciona en listas igual que en strings: extrae un trozo con la sintaxis `p[i:j]`, desde la posición `i` hasta la `j-1`. El límite superior no se incluye.

También podemos omitir cualquiera de los dos extremos.

```python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

print(notas[1:4])    # [5.8, 4.2, 3.9]
print(notas[:3])     # [6.5, 5.8, 4.2]
print(notas[2:])     # [4.2, 3.9, 6.1]
print(notas[-2:])    # [3.9, 6.1]
print(notas[:])      # copia completa
```

## LIS-06 · Pegar: `+`

El operador `+` une dos listas en una sola, en el orden en que las escribimos. Crea una lista nueva: no modifica las originales.

```python
grupo_a = ["Ana", "Luis"]
grupo_b = ["María", "Pedro", "Diego"]

curso = grupo_a + grupo_b
print(curso)
# ['Ana', 'Luis', 'María', 'Pedro', 'Diego']

print(grupo_a)   # grupo_a NO cambió
print(grupo_b)   # grupo_b tampoco
```

Combinando slicing y pegado podemos crear listas nuevas a partir de trozos:

```python
# Insertar un elemento al medio
numeros = [1, 2, 3, 4, 5]
con_cero = numeros[:2] + [0] + numeros[2:]
print(con_cero)   # [1, 2, 0, 3, 4, 5]
```

## LIS-09 · Extraer: `.pop()`

El método `.pop()` hace dos cosas a la vez: **quita** el último elemento de la lista y lo **devuelve** como resultado. Suele usarse con una asignación.

**Cuidado:** si la lista está vacía, `.pop()` lanza un error. Siempre verificar `len(lista) > 0` antes de llamarlo.

```python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

ultima = notas.pop()

print("Elemento extraído:", ultima)   # 6.1
print("Lista actualizada:", notas)    # [6.5, 5.8, 4.2, 3.9]
```

```python
# Patrón: vaciar una lista con while
pendientes = ["estudiar", "ejercicio", "compras"]

while len(pendientes) > 0:
    tarea = pendientes.pop()
    print("Haciendo:", tarea)

print("Lista final:", pendientes)
```

```python
def mi_pop(lista, indice=-1):
    # Si la lista está vacía, no podemos extraer nada
    if len(lista) == 0:
        raise IndexError("pop de una lista vacía")

    # Manejar índices negativos
    if indice < 0:
        indice = len(lista) + indice

    # Validar que el índice esté dentro del rango correcto
    if indice < 0 or indice >= len(lista):
        raise IndexError("índice fuera de rango")

    # Guardar el elemento que vamos a extraer
    elemento = lista[indice]

    # Reconstruir la lista original sin el elemento extraído usando slicing
    # Esto modifica la lista de manera directa (mutabilidad)
    lista[:] = lista[:indice] + lista[indice + 1:]

    # Retornar el elemento extraído
    return elemento

# Ejemplo de uso:
mi_lista = [10, 20, 30, 40, 50]

# 1. Pop sin índice (extrae el último)
ultimo = mi_pop(mi_lista)
print("Elemento extraído:", ultimo) # 50
print("Lista modificada:", mi_lista) # [10, 20, 30, 40]

# 2. Pop con un índice específico (extrae el de la posición 1)
segundo = mi_pop(mi_lista, 1)
print("Elemento extraído:", segundo) # 20
print("Lista modificada:", mi_lista) # [10, 30, 40]
```
