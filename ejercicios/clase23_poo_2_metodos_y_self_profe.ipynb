## OBJ-08 · Un método puede llamar a otro

Dentro de un método, para llamar a OTRO método del mismo objeto, se usa `self.otro_metodo()`.

Esto es muy útil para no repetir lógica.

```python
class Carta:
    def __init__(self, numero, pinta):
        self.numero = numero
        self.pinta = pinta

    def __str__(self):
        return str(self.numero) + " de " + self.pinta

    def es_figura(self):
        return self.numero >= 11

    def valor_blackjack(self):
        if self.numero == 1:
            return 11
        elif self.es_figura():        # usando otro método propio
            return 10
        else:
            return self.numero

# Pruebas
cartas = [Carta(1, "♥"), Carta(5, "♣"), Carta(11, "♦"), Carta(13, "♠")]
for c in cartas:
    print(c, "vale en blackjack:", c.valor_blackjack())
```

## Ejemplo integrador · clase Estudiante completa

Esta clase junta todo lo visto hoy: `__init__`, `__str__`, varios métodos, uno que llama a otro con `self`.

```python
class Estudiante:
    def __init__(self, nombre, carrera):
        self.nombre = nombre
        self.carrera = carrera
        self.notas = []

    def agregar_nota(self, nota):
        self.notas.append(nota)

    def promedio(self):
        if len(self.notas) == 0:
            return 0
        return sum(self.notas) / len(self.notas)

    def esta_aprobado(self):
        return self.promedio() >= 4.0    # usa otro método

    def __str__(self):
        return self.nombre + " (promedio: " + str(round(self.promedio(), 2)) + ")"

# Uso
ana = Estudiante("Ana", "Ingeniería")
ana.agregar_nota(6.5)
ana.agregar_nota(5.0)
ana.agregar_nota(7.0)

print(ana)
print("¿Aprobada?", ana.esta_aprobado())

luis = Estudiante("Luis", "Letras")
luis.agregar_nota(3.5)
luis.agregar_nota(3.8)
print(luis)
print("¿Aprobado?", luis.esta_aprobado())
```
