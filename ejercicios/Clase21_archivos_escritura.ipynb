## ARC-04 · Escribir un archivo .txt

Tres pasos: `open("w")`, `print(..., file=archivo)`, `close()`.

**Cuidado:** el modo `"w"` sobrescribe el archivo si ya existe. Si quieres agregar al final en vez de sobrescribir, usa `"a"` (append).

````python
# Escribir un reporte simple
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

archivo = open("reporte.txt", "w")
print("Reporte de notas", file=archivo)
print("----------------", file=archivo)

for i in range(len(notas)):
    print("Nota", i + 1, ":", notas[i], file=archivo)

suma = 0
for n in notas:
    suma = suma + n
print("Promedio:", suma / len(notas), file=archivo)

archivo.close()

# Verificar: leer y mostrar lo escrito
archivo = open("reporte.txt", "r")
print(archivo.read())
archivo.close()
````

## Patrón · leer, procesar, escribir

Combinando todo: leer un archivo, calcular algo y escribir un archivo con los resultados.

````python
# Leer notas.csv, calcular promedios y escribir promedios.csv
entrada = open("notas.csv", "r")
lineas = entrada.readlines()
entrada.close()

salida = open("promedios.csv", "w")
for linea in lineas:
    partes = linea.strip().split(",")
    nombre = partes[0]
    nota1 = float(partes[1])
    nota2 = float(partes[2])
    nota3 = float(partes[3])
    prom = (nota1 + nota2 + nota3) / 3
    print(nombre, round(prom, 2), sep=",", file=salida)
salida.close()

# Verificar
archivo = open("promedios.csv", "r")
print(archivo.read())
archivo.close()
````

# Taller de debugging · bugs sutiles

Hoy los bugs son más sutiles: no explotan con un error claro, simplemente dan resultados incorrectos. Hay que leer con cuidado y pensar en qué debería hacer el código antes de ejecutarlo.

### Bug 1 · promedio mal calculado

````python
# Calcular el promedio de los números positivos
numeros = [3, -1, 5, -4, 7, 2, -8]
suma = 0
cantidad = 0

for n in numeros:
    if n > 0:
        suma = suma + n
        cantidad = cantidad + 1

promedio = suma / len(numeros)
print("Promedio de positivos:", promedio)
# Esperado: 17 / 4 = 4.25
# Lo que da: 17 / 7 = 2.43
````

### Bug 2 · lógica compuesta mal armada

````python
# Debería aprobar si nota >= 4.0 Y asistencia > 70
nota = 3.0
asistencia = 90

if nota >= 4.0 or asistencia > 70:
    print("Aprobado")
else:
    print("Reprobado")

# Con nota=3.0 y asistencia=90, debería decir Reprobado (nota baja)
# pero dice Aprobado
````

### Bug 3 · modificar lista mientras se itera

````python
# Quitar los números negativos de la lista
numeros = [1, -2, 3, -4, 5, -6, 7]

for n in numeros:
    if n < 0:
        numeros.remove(n)

print(numeros)
# Esperado: [1, 3, 5, 7]
# Lo que da: [1, 3, -4, 5, 7]  (-4 no fue eliminado!)
````

### Bug 4 · acumulador no reseteado

````python
# Calcular el promedio de cada fila de una matriz
matriz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

suma = 0
for fila in matriz:
    for valor in fila:
        suma = suma + valor
    promedio = suma / len(fila)
    print("Promedio de la fila:", promedio)

# Esperado: 2.0, 5.0, 8.0
# Lo que da: 2.0, 7.0, 15.0
````

### Bug 5 · recursividad con un caso base incompleto

````python
def suma_lista(lista):
    if lista == []:
        return 0
    else:
        return lista[0] + suma_lista(lista[1])

print(suma_lista([1, 2, 3, 4]))
# Esperado: 10
````

# Desafío final · pipeline con 4 bugs mezclados

El siguiente programa debería:

1. Leer `notas.csv` y `asistencia.csv`.
2. Para cada estudiante, calcular su promedio.
3. Determinar si está **aprobado**: promedio >= 4.0 **Y** asistencia >= 70.
4. Escribir `informe.csv` con cuatro columnas: `nombre, promedio, asistencia, estado`.
5. Imprimir cuántos aprobaron y cuántos reprobaron.

Pero tiene **4 bugs mezclados**. Encuéntralos todos para que funcione correctamente.

````python
# Leer notas
arch_notas = open("notas.csv", "r")
lineas_notas = arch_notas.readlines()
arch_notas.close()

# Leer asistencia
arch_asis = open("asistencia.csv", "r")
lineas_asis = arch_asis.readlines()
arch_asis.close()

# Procesar notas
nombres = []
promedios = []
for linea in lineas_notas:
    partes = linea.split(",")
    nombres.append(partes[0])
    nota1 = float(partes[1])
    nota2 = float(partes[2])
    nota3 = float(partes[3])
    promedios.append((nota1 + nota2 + nota3) / 3)

# Procesar asistencia
asistencias = []
for linea in lineas_asis:
    partes = linea.strip().split(",")
    asistencias.append(partes[1])

# Combinar y escribir informe
salida = open("informe.csv", "w")
aprobados = 0
reprobados = 0

for i in range(len(nombres)):
    nombre = nombres[i]
    prom = promedios[i]
    asis = asistencias[i]

    if prom >= 4.0 or asis >= 70:
        estado = "aprobado"
        aprobados = aprobados + 1
    else:
        estado = "reprobado"
        reprobados = reprobados + 1

    print(nombre, prom, asis, estado, sep=",", file=salida)

print("Aprobados:", aprobados)
print("Reprobados:", reprobados)
````
