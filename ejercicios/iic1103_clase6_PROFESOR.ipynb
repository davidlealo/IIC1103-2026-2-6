## FLU-05 · `for … in range`

Se usa cuando sabemos de antemano cuántas veces queremos repetir.

```
for i in range(n):
    # bloque que se ejecuta n veces
```

`range(n)` genera los números `0, 1, 2, …, n-1`. En cada vuelta la variable `i` toma uno de esos valores.

````python
list(range(10))
````

````python
list(range(11))
````

````python
list(range(1, 11))
````

````python
list(range(0, 10, 2))
````

````python
list(range(0, 11, 2))
````

````python
list(range(0))
````

````python
# ciclo for de numeros pares
for i in range(0, 11, 2):
    print(i)
````

````python
# Ejemplo: repetir un mensaje 5 veces
for i in range(5):
    print("Vuelta", i)

print("Fin")
````

````python
fruits = ["apple", "banana", "cherry"]

for x in fruits:
  print(x)
````

````python
frutas = ["manzana", "plátano", "pera"]
for fruta in frutas:
  print("El nombre de la fruta es: ", fruta)
````

````python
# Enumerar las frutas
frutas = ["manzana", "plátano", "pera"]

for i, fruta in enumerate(frutas):
  print(i, fruta)
````

````python
# Enumerar las frutas el número va a partir en 1
frutas = ["manzana", "plátano", "pera"]

for i, fruta in enumerate(frutas, 1):
  print(i, fruta)
````

`range` tiene tres formas:

| Forma | Qué genera | Ejemplo |
|---|---|---|
| `range(n)` | 0, 1, …, n-1 | `range(5)` → 0, 1, 2, 3, 4 |
| `range(a, b)` | a, a+1, …, b-1 | `range(2, 6)` → 2, 3, 4, 5 |
| `range(a, b, paso)` | de a a b saltando de paso en paso | `range(0, 10, 2)` → 0, 2, 4, 6, 8 |
| `range(b, a, -1)` | de b hacia abajo hasta a+1 | `range(5, 0, -1)` → 5, 4, 3, 2, 1 |

En todos los casos el límite superior nunca se incluye.

````python
# Ejemplo: tabla del 5
for i in range(1, 11):
    resultado = 5 * i
    print("5 x", i, "=", resultado)
````

## FLU-06 · `while`

Se usa cuando no sabemos cuántas veces vamos a repetir: sólo sabemos cuándo parar.

```
while condicion:
    # bloque que se repite mientras condicion sea True
```

El ciclo evalúa la condición, ejecuta el bloque si es `True`, y vuelve a evaluar. Termina cuando la condición se hace `False`.

Un `while` bien escrito tiene tres piezas:

1. **Valor inicial** de la variable de control
2. **Condición** para seguir repitiendo
3. **Actualización** de la variable dentro del bloque

Si falta la actualización, el ciclo puede quedar corriendo para siempre (ciclo infinito).

````python
# Ejemplo: contador del 1 al 5
i = 1
while i <= 5:
    print("Iteración", i)
    i = i + 1

print("Fin del ciclo")
````

Un uso muy común del `while` es validar la entrada del usuario: si ingresó algo inválido, se le pide de nuevo.

````python
# Ejemplo: pedir la edad hasta que sea válida
edad = int(input("Ingresa tu edad: "))

while edad < 0:
    print("La edad no puede ser negativa")
    edad = int(input("Ingresa tu edad: "))

print("Edad registrada:", edad)
````

## FLU-07 · Iteraciones dentro de iteraciones

Podemos poner un ciclo dentro de otro. Por cada vuelta del ciclo externo, el ciclo interno se ejecuta completo.

```
for i in range(3):
    for j in range(2):
        print(i, j)
```

Si el externo hace 3 vueltas y el interno hace 2, el bloque interno se ejecuta 3 × 2 = 6 veces en total.

La indentación indica qué instrucciones pertenecen a cada ciclo. También podemos anidar `while` con `for`, o cualquier combinación.

````python
# Ejemplo: recorrer una grilla
for i in range(3):
    for j in range(2):
        print("i =", i, ", j =", j)
````

````python
# Ejemplo: tabla de multiplicar del 1 al 3
for tabla in range(1, 4):
    print("Tabla del", tabla)
    for i in range(1, 6):
        print(tabla, "x", i, "=", tabla * i)
    print()
````
