## STR-04 · String slicing

El slicing permite extraer un trozo de un string con la sintaxis:

```
s[i:j]
```

Entrega los caracteres desde la posición `i` hasta la `j-1` (no incluye `j`, igual que `range`).

También se pueden omitir los extremos:

| Forma | Qué entrega |
|---|---|
| `s[:j]` | desde el inicio hasta `j-1` |
| `s[i:]` | desde `i` hasta el final |
| `s[:]` | copia completa |
| `s[-k:]` | los últimos `k` caracteres |
| `s[::k]` | cada `k` caracteres |
| `s[::-1]` | el string invertido |

````python
# Ejemplos de slicing
palabra = "Python"

print(palabra[1:4])   # yth
print(palabra[:3])    # Pyt
print(palabra[2:])    # thon
print(palabra[-3:])   # hon
print(palabra[::2])   # Pto
print(palabra[::-1])  # nohtyP
````

## STR-06 · Inmutabilidad de los strings

Un string **no se puede modificar** después de creado. Si intentas asignarle un carácter nuevo a una posición, Python lanza un error:

```python
palabra = "casa"
palabra[0] = "m"   # TypeError
```

Para "modificar" un string, construimos uno nuevo usando slicing (STR-04) y concatenación (STR-05), y lo guardamos en una variable.

````python
# Esto da error - descomentar para verlo
# palabra = "casa"
# palabra[0] = "m"
````

````python
# Patrón: construir uno nuevo
palabra = "casa"
nueva = "m" + palabra[1:]
print(nueva)     # masa
print(palabra)   # casa  (el original no cambió)

# Si queremos reemplazar la variable, hay que reasignar:
palabra = "m" + palabra[1:]
print(palabra)   # masa
````

## STR-07 · ¿Es dígito un carácter?

Para saber si un carácter es un dígito usamos el operador `in` (STR-03) comparando contra los diez caracteres posibles:

```python
if c in "0123456789":
    # c es un dígito
```

````python
# Contar cuántos caracteres son dígitos
texto = "casa123numero456"
digitos = 0

for c in texto:
    if c in "0123456789":
        digitos = digitos + 1

print("Dígitos encontrados:", digitos)   # 6
````

## STR-08 · Métodos de strings

Los strings tienen **métodos**: funciones que se llaman con un punto después del string.

| Método | Qué hace |
|---|---|
| `s.lower()` | Devuelve el string en minúsculas |
| `s.upper()` | Devuelve el string en mayúsculas |
| `s.islower()` | `True` si todas las letras son minúsculas |
| `s.isupper()` | `True` si todas las letras son mayúsculas |

**Importante:** `lower()` y `upper()` no modifican el string original (recuerda la inmutabilidad); devuelven uno nuevo que hay que guardar.

````python
# lower() y upper() devuelven un string nuevo
palabra = "Hola"

print(palabra.upper())   # HOLA
print(palabra)           # Hola  (sigue igual)

# Para reemplazar, hay que reasignar
palabra = palabra.upper()
print(palabra)           # HOLA
````

````python
# islower() y isupper() devuelven booleanos
print("hola".islower())   # True
print("Hola".islower())   # False (la H es mayúscula)
print("HOLA".isupper())   # True
````

## STR-09 · Buscar el substring más largo con cierta propiedad

Un patrón común: recorrer el string buscando la **secuencia más larga de caracteres consecutivos** que cumplen una condición.

**Patrón general** con dos variables auxiliares:

- `mejor`: el substring más largo encontrado hasta ahora.
- `actual`: el substring que estamos armando en este momento.

Reglas:

1. Si el carácter actual cumple la condición, lo agregamos a `actual`.
2. Si no cumple, reseteamos `actual` a `""`.
3. Cada vez que `actual` crece, lo comparamos con `mejor` y actualizamos si es más largo.

````python
# Secuencia más larga de dígitos consecutivos
texto = "abc12fg34567hi89"
mejor  = ""
actual = ""

for c in texto:
    if c in "0123456789":
        actual = actual + c
        if len(actual) > len(mejor):
            mejor = actual
    else:
        actual = ""

print("Secuencia más larga:", mejor)   # 34567
````
