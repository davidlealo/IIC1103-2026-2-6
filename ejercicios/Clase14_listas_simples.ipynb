## LIS-02 · Longitud y acceso

Dos operaciones que ya conocíamos de strings:

- `len(p)` devuelve el número de elementos de la lista.
- `p[i]` devuelve el elemento en la posición `i`.

Los índices empiezan en **0**: el primer elemento es `p[0]` y el último es `p[len(p) - 1]`. También funcionan los índices negativos: `p[-1]` es el último.

````python
# Medir y acceder
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

print("Longitud:", len(notas))
print("Primera nota:", notas[0])
print("Tercera nota:", notas[2])
print("Última nota:", notas[-1])
print("Última nota (otra forma):", notas[len(notas) - 1])
````

## LIS-03 · Iterar elemento a elemento

Para procesar todos los elementos de una lista tenemos tres formas, exactamente iguales a las que vimos para strings.

### Forma 1 · `for e in p`

La más simple y directa. En cada vuelta, la variable toma el valor de un elemento.

````python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

for n in notas:
    print(n)
````

### Forma 2 · `for i in range(len(p))`

Recorre las posiciones. Útil cuando necesitamos saber en qué índice estamos.

````python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]

for i in range(len(notas)):
    print("Nota", i, "->", notas[i])
````

### Forma 3 · `while`

La más flexible. Útil cuando queremos detener el recorrido antes de llegar al final.

````python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]
i = 0

while i < len(notas):
    print(notas[i])
    i = i + 1
````

### ¿Cuál usar?

| Forma | Cuándo |
|---|---|
| `for e in p` | Sólo importa el valor |
| `for i in range(len(p))` | Necesitamos la posición |
| `while` | Parada anticipada o avance flexible |

### Ejemplo: calcular el promedio

````python
notas = [6.5, 5.8, 4.2, 3.9, 6.1]
suma = 0

for n in notas:
    suma = suma + n

promedio = suma / len(notas)
print("Promedio:", promedio)
````

## LIS-04 · Pertenencia con `in`

El operador `in` responde a *"¿este elemento está en la lista?"*. Devuelve `True` o `False`, así que se puede usar directamente en un `if`.

Importante: compara el elemento **completo**, no partes. Y es sensible al tipo: `5 in ["5"]` es `False` porque un entero no es igual a un string.

````python
nombres = ["Ana", "Luis", "María", "Pedro"]

print("Ana" in nombres)      # True
print("ana" in nombres)      # False (sensible a mayúsculas)
print("Diego" in nombres)    # False

notas = [6.5, 5.8, 4.2]
print(4.2 in notas)          # True
print(7.0 in notas)          # False
````
