## 4. Operaciones con números

| Operador | Nombre |
|----------|--------|
| `+` | suma |
| `-` | resta |
| `*` | multiplicación |
| `/` | división |
| `//` | división entera |
| `%` | módulo (resto) |
| `**` | potencia |

Precedencia: primero `**`, después `*`, `/`, `//`, `%`, al final `+` y `-`. Los paréntesis alteran el orden.

```python
print(3 + 4 * 2)      # 11
print((3 + 4) * 2)    # 14
print(7 / 2)          # 3.5
print(7 // 2)         # 3
print(7 % 2)          # 1
print(2 ** 10)        # 1024
```

## 5. Comparaciones

Devuelven `True` o `False`.

| Operador | Significado |
|----------|-------------|
| `>` `<` | mayor / menor |
| `>=` `<=` | mayor o igual / menor o igual |
| `==` | igual |
| `!=` | distinto |

Un `=` asigna. Dos `==` comparan. No es lo mismo.

```python
print(5 > 3)     # True
print(5 == 5)    # True
print(5 != 5)    # False
print(5 >= 5)    # True
```
