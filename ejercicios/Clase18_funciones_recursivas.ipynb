Una función es **recursiva** cuando se llama a sí misma. Es una forma natural de resolver problemas que se pueden descomponer en versiones más pequeñas del mismo problema. Cuatro puntos:

- **REC-01** · Concepto: caso base y caso recursivo

Toda función recursiva tiene **dos partes**:

1. **Caso base:** el problema más simple, que se resuelve directamente.
2. **Caso recursivo:** el problema general se resuelve usando el resultado del mismo problema para un caso más chico.

## REC-01 · Primer ejemplo: cuenta regresiva

Antes de entrar a cálculos, un ejemplo que sólo imprime: contar hacia atrás desde `n` hasta 0.

````python
def cuenta_regresiva(n):
    if n == 0:
        print("¡Fin!")
    else:
        print(n)
        cuenta_regresiva(n - 1)

cuenta_regresiva(5)
````

Las dos partes claramente identificables:

- **Caso base:** `n == 0`, imprime "¡Fin!" y no se llama a sí misma.
- **Caso recursivo:** imprime `n` y se llama con `n - 1` (más cerca del caso base).

## REC-02 · Recursividad con números: factorial

El factorial de `n` es `n! = n × (n-1) × ... × 1`, con `0! = 1` por convención.

La observación clave: `n! = n × (n-1)!`. Si confiamos en que `(n-1)!` está resuelto, basta con multiplicar por `n`.

````python
def factorial(n):
    if n == 0:
        return 1
    else:
        return n * factorial(n - 1)

print(factorial(0))   # 1
print(factorial(1))   # 1
print(factorial(4))   # 24
print(factorial(5))   # 120
````

### Fibonacci

Otro clásico: la sucesión 0, 1, 1, 2, 3, 5, 8, 13, ... donde cada número es la suma de los dos anteriores. Esta vez tenemos **dos casos base**.

````python
def fibonacci(n):
    if n == 0:
        return 0
    elif n == 1:
        return 1
    else:
        return fibonacci(n - 1) + fibonacci(n - 2)

for i in range(10):
    print("fib(", i, ") =", fibonacci(i))
````

## REC-03 · Recursividad con strings

La misma estructura, ahora sobre strings. El **caso base** es el string vacío `""`. El **avance** es trabajar con el string sin su primera letra: `s[1:]`.

### Longitud de un string

````python
def largo(s):
    if s == "":
        return 0
    else:
        return 1 + largo(s[1:])

print(largo(""))         # 0
print(largo("hola"))     # 4
print(largo("Python"))   # 6
````

### Reverso de un string

**Idea:** el reverso de `"hola"` es el reverso de `"ola"` seguido de `"h"`.

````python
def reverso(s):
    if s == "":
        return ""
    else:
        return reverso(s[1:]) + s[0]

print(reverso("hola"))     # 'aloh'
print(reverso("python"))   # 'nohtyp'
````

## REC-04 · Recursividad con listas

Misma estructura una vez más. **Caso base:** lista vacía `[]`. **Avance:** trabajar con la lista sin su primer elemento, `lista[1:]`.

### Suma de los elementos

````python
def suma(lista):
    if lista == []:
        return 0
    else:
        return lista[0] + suma(lista[1:])

print(suma([]))                # 0
print(suma([1, 2, 3]))         # 6
print(suma([10, 20, 30, 40]))  # 100
````

### Máximo de una lista

El máximo es el mayor entre el primer elemento y el máximo del resto. Caso base: la lista tiene un solo elemento.

````python
def maximo(lista):
    if len(lista) == 1:
        return lista[0]
    else:
        resto = maximo(lista[1:])
        if lista[0] > resto:
            return lista[0]
        else:
            return resto

print(maximo([3]))             # 3
print(maximo([1, 5, 2, 8, 3])) # 8
print(maximo([10, 9, 8, 7]))   # 10
````
