Ejecuta la siguiente celda. Observa qué pasa cuando intentas imprimir el objeto directamente.

Python muestra algo como `<__main__.Carta object at 0x...>`, que no sirve a nadie. Agreguemos un método `__str__` que devuelva un string legible.

```python
class Carta:
    def __init__(self, numero, pinta):
        self.numero = numero
        self.pinta = pinta

    def __str__(self):
        return str(self.numero) + " de " + self.pinta

as_corazones = Carta(1, "corazones")
rey_picas = Carta(13, "picas")

print(as_corazones)
print(rey_picas)
```

## OBJ-07 · Definir métodos propios

Los métodos son funciones que viven dentro de la clase. Siempre reciben `self` como primer parámetro y pueden acceder a los atributos del objeto.

```python
class Carta:
    def __init__(self, numero, pinta):
        self.numero = numero
        self.pinta = pinta

    def __str__(self):
        return str(self.numero) + " de " + self.pinta

    def es_figura(self):
        return self.numero >= 11

    def valor(self):
        if self.numero == 1:
            return 11       # as vale 11
        else:
            return self.numero

# Pruebas
rey_picas = Carta(13, "picas")
siete_treboles = Carta(7, "tréboles")
as_corazones = Carta(1, "corazones")

print(rey_picas, "- es figura?", rey_picas.es_figura())
print(siete_treboles, "- es figura?", siete_treboles.es_figura())
print(as_corazones, "- vale", as_corazones.valor())
```

**Observa:**
- `rey_picas.es_figura()` NO lleva `self` entre los paréntesis. Python lo pasa automáticamente.
- Dentro del método, `self.numero` accede al atributo `numero` del objeto que llamó al método.

## OBJ-08 · Un método puede llamar a otro

Dentro de un método, para llamar a OTRO método del mismo objeto, se usa `self.otro_metodo()`.

Esto es muy útil para no repetir lógica.

```python
class Carta:
    def __init__(self, numero, pinta):
        self.numero = numero
        self.pinta = pinta

    def __str__(self):
        return str(self.numero) + " de " + self.pinta

    def es_figura(self):
        return self.numero >= 11

    def valor_blackjack(self):
        if self.numero == 1:
            return 11
        elif self.es_figura():        # usando otro método propio
            return 10
        else:
            return self.numero

# Pruebas
cartas = [Carta(1, "♥"), Carta(5, "♣"), Carta(11, "♦"), Carta(13, "♠")]
for c in cartas:
    print(c, "vale en blackjack:", c.valor_blackjack())
```

## Ejemplo integrador · clase Estudiante completa

Esta clase junta todo lo visto hoy: `__init__`, `__str__`, varios métodos, uno que llama a otro con `self`.

```python
class Estudiante:
    def __init__(self, nombre, carrera):
        self.nombre = nombre
        self.carrera = carrera
        self.notas = []

    def agregar_nota(self, nota):
        self.notas.append(nota)

    def promedio(self):
        if len(self.notas) == 0:
            return 0
        return sum(self.notas) / len(self.notas)

    def esta_aprobado(self):
        return self.promedio() >= 4.0    # usa otro método

    def __str__(self):
        return self.nombre + " (promedio: " + str(round(self.promedio(), 2)) + ")"

# Uso
ana = Estudiante("Ana", "Ingeniería")
ana.agregar_nota(6.5)
ana.agregar_nota(5.0)
ana.agregar_nota(7.0)

print(ana)
print("¿Aprobada?", ana.esta_aprobado())

luis = Estudiante("Luis", "Letras")
luis.agregar_nota(3.5)
luis.agregar_nota(3.8)
print(luis)
print("¿Aprobado?", luis.esta_aprobado())
```
