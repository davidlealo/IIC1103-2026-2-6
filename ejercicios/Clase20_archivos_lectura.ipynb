## ARC-02 · Leer un archivo .txt

Tres pasos: `open`, `readlines`, `close`. Después se recorre la lista de líneas.

**Importante:** `readlines()` devuelve strings con el `\n` al final. Casi siempre hay que hacer `strip()`.

```python
# Leer nombres.txt y saludar a cada uno
archivo = open("nombres.txt", "r")
lineas = archivo.readlines()
archivo.close()

for linea in lineas:
    nombre = linea.strip()
    print("Hola,", nombre)
```

## ARC-03 · Leer un archivo .csv

Mismo patrón que .txt, pero cada línea tiene varios campos separados por coma. Se usa `.split(",")` sobre la línea ya "limpia" (con strip).

**Importante:** todo viene como string. Si queremos números, hay que convertir con `float()` o `int()`.

```python
# Leer notas.csv e imprimir el promedio de cada estudiante
archivo = open("notas.csv", "r")
lineas = archivo.readlines()
archivo.close()

for linea in lineas:
    partes = linea.strip().split(",")
    nombre = partes[0]
    nota1 = float(partes[1])
    nota2 = float(partes[2])
    nota3 = float(partes[3])
    promedio = (nota1 + nota2 + nota3) / 3
    print(nombre, "-> promedio:", round(promedio, 2))
```

# Desafío · programa con 3 bugs mezclados

El siguiente programa debería:
1. Leer `notas.csv`.
2. Guardar los nombres de todos los aprobados (promedio >= 4.0) en una lista.
3. Guardar los promedios en otra lista.
4. Imprimir cuántos aprobaron, quién tuvo el mejor promedio y cuál fue el peor promedio.

Pero tiene **3 bugs sutiles**. Encuéntralos todos para que funcione correctamente.

```python
archivo = open("notas.csv", "r")
lineas = archivo.readlines()
archivo.close()

aprobados = []
promedios = []

for linea in lineas:
    partes = linea.split(",")
    nombre = partes[0]
    nota1 = partes[1]
    nota2 = float(partes[2])
    nota3 = float(partes[3])
    promedio = (nota1 + nota2 + nota3) / 3
    if promedio >= 4.0:
        aprobados.append(nombre)
        promedios.append(promedio)

print("Cantidad de aprobados:", len(aprobados))

mejor = promedios[0]
peor = promedios[0]
nombre_mejor = aprobados[0]

for i in range(len(promedios)):
    if promedios[i] > mejor:
        mejor = promedios[i]
        nombre_mejor = aprobados[i]
    if promedios[i] > peor:
        peor = promedios[i]

print("Mejor promedio:", nombre_mejor, "con", mejor)
print("Peor promedio:", peor)
```

```python
# =====================================================
# EJERCICIO DE DEBUGGING - Clase 20
# Analizador de inventario de una librería
# =====================================================
#
# Este programa debería leer libros.csv y producir:
#
#   1. El valor total del inventario (precio x stock de cada libro).
#   2. El libro más caro de la librería.
#   3. La cantidad de libros agotados (stock == 0).
#   4. El precio promedio de los libros de la categoría "Novela".
#
# Tiene 4 bugs mezclados. Encuéntralos todos y corrígelos.
# =====================================================


archivo = open("libros.csv", "r")
lineas = archivo.readlines()
archivo.close()


# --- 1. Valor total del inventario ----------------------
valor_total = 0
for linea in lineas:
    partes = linea.strip().split(",")
    precio = partes[2]
    stock = int(partes[3])
    valor_total = valor_total + precio * stock

print("Valor total del inventario: $", valor_total)


# --- 2. Libro más caro ----------------------------------
mas_caro_titulo = ""
mas_caro_precio = 0

for linea in lineas:
    partes = linea.strip().split(",")
    titulo = partes[0]
    precio = int(partes[2])
    if precio < mas_caro_precio:
        mas_caro_precio = precio
        mas_caro_titulo = titulo

print("Libro más caro:", mas_caro_titulo, "- $", mas_caro_precio)


# --- 3. Libros agotados ---------------------------------
agotados = 0
for linea in lineas:
    partes = linea.strip().split(",")
    stock = int(partes[3])
    if stock == 0:
        agotados = agotados + 1

print("Libros agotados:", agotados)


# --- 4. Precio promedio de los libros de categoría "Novela"
suma = 0
cantidad = 0
for linea in lineas:
    partes = linea.split(",")
    categoria = partes[4]
    precio = int(partes[2])
    if categoria == "Novela":
        suma = suma + precio
        cantidad = cantidad + 1

promedio = suma / cantidad
print("Precio promedio de Novelas: $", promedio)
```
