## OBJ-01 · Ustedes ya han usado objetos

En Python, prácticamente todo es un objeto. Los strings, las listas y los números tienen sus propios **métodos** (acciones que pueden hacer).

Ejecuta las celdas siguientes para recordar.

````python
# Un string es un objeto con sus propios métodos
nombre = "Ana"
print(nombre.upper())
print(nombre.lower())
print(len(nombre))
````

````python
# Una lista también
notas = [6.5, 5.0, 7.0]
notas.append(4.5)
print(notas)
print(len(notas))
````

La sintaxis `objeto.metodo()` o `objeto.atributo` que ya conocen es la base de POO. Hoy vamos a crear nuestros propios tipos de objetos.

## OBJ-02 y OBJ-03 · Crear objetos y acceder a sus atributos

Supongamos que alguien ya definió la clase `Carta`. Para crear una carta y leer sus atributos:

````python
# Definición preexistente (ya la veremos en detalle en OBJ-05)
class Carta:
    def __init__(self, numero, pinta):
        self.numero = numero
        self.pinta = pinta

# Crear objetos
as_corazones = Carta(1, "corazones")
rey_picas = Carta(13, "picas")
dama_treboles = Carta(12, "tréboles")

# Acceder a atributos
print(as_corazones.numero)
print(as_corazones.pinta)
print(rey_picas.numero, "de", rey_picas.pinta)
````

## OBJ-04 y OBJ-05 · Definir una clase con `__init__`

### Estructura básica

```python
class NombreClase:
    def __init__(self, parametro1, parametro2):
        self.atributo1 = parametro1
        self.atributo2 = parametro2
```

### Tres piezas clave

- **`class NombreClase:`** · declara la clase. Convención: empieza con mayúscula.
- **`def __init__(self, ...):`** · es el constructor. Define qué recibe un objeto nuevo.
- **`self.atributo = valor`** · guarda el valor como atributo del objeto.

El parámetro `self` siempre va primero y representa al objeto que se está creando. Python lo pasa automáticamente.

````python
# Ejemplo completo: clase Estudiante
class Estudiante:
    def __init__(self, nombre, edad, carrera):
        self.nombre = nombre
        self.edad = edad
        self.carrera = carrera
        self.notas = []    # inicializado vacío, no viene como parámetro

# Crear objetos
ana = Estudiante("Ana", 20, "Ingeniería")
luis = Estudiante("Luis", 22, "Letras")

print(ana.nombre, "tiene", ana.edad, "años y estudia", ana.carrera)
print(luis.nombre, "tiene", luis.edad, "años y estudia", luis.carrera)
print("Notas de Ana:", ana.notas)
````

**Observa:** no todos los atributos tienen que venir como parámetros. `notas` se inicializa como una lista vacía que se podrá llenar después.
